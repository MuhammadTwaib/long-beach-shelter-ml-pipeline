```python
import sys, pandas as pd
sys.path.append('..')
from src import config
raw = pd.read_csv(config.RAW_FILE, dtype=str, keep_default_na=False, encoding='utf-8-sig')
raw.shape
```

```python
# empty strings per column = missing values
(raw.apply(lambda s: s.str.strip().eq('')).sum()).sort_values(ascending=False)
```

```python
# spotting inconsistent categories (double spaces, typos)
raw['Intake Condition'].value_counts()
```

```python
# DOB after intake date?
d = lambda c: pd.to_datetime(raw[c], errors='coerce')
(d('DOB') > d('Intake Date')).sum()
```

```python
# same animal, same intake date = duplicate stay
raw.duplicated(['Animal ID', 'Intake Date']).sum()
```
